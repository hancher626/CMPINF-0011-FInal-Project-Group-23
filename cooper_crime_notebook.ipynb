Looking at crime data in Pittsburgh...

Let's see what we're working with - what does this data look like?

```python
import pandas as pd

df = pd.read_csv("data/pittsburgh_police_data.csv")

print(df.shape)
print(df.columns)
df.head(5)
```

59353 entries, not bad. I noticed there was a neighborhood attribute, let's look at that next...

```python
print(df["Neighborhood"].values)
print(df["Neighborhood"].value_counts())
```

Some neighborhoods are already looking better than others- Bluff doesn't look great while St. Clair has only 17 incidents in the past 2.5 years. I also want to look at crime classifications - how common are felonies vs misdemeanors vs infractions?

```python
print(df["ArrestCharge_Felony_Misdemeanor_Description"].value_counts())
```

Mostly misdemeanors and infractions, fewer felonies. Last thing for this preliminary analysis- can we group these by neighborhood?

```python
df.groupby("Neighborhood")["ArrestCharge_Felony_Misdemeanor_Description"].value_counts()
```

Yikes, Windgap has more felonies than misdemeanors... not many crimes total though.
